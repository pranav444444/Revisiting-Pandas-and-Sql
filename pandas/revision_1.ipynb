## Data-Filtration

```python
#give me data about only those regions which belong to south
filtered_condition=df['region']=='South'
df.loc[filtered_condition]
```

```python
df['region']=='South'
```

```python
df.loc[df['region']=='South']
```

```python
#filter roes based on region='South' and show only order_id and order_date as columns
df.loc[df['region']=='South',['order_id','order_date']]
```

```python
#Multiple Filter conditons
df.loc[(df['region']=='South') & (df['profit']>0)]
```

```python
#we can ols use OR operation by using '|'
df.loc[(df['region']=='South') | (df['profit']>0)]
```

```python
#without using loc we can also use filter condition
df[df['region']=='South']
```

```python
#inverse the condition
#give me all regions except South
df[~(df['region']=='South')]
```
