```python
df.loc[df['region']=='South']
```

```python
#filter roes based on region='South' and show only order_id and order_date as columns
df.loc[df['region']=='South',['order_id','order_date']]
```

```python
#Multiple Filter conditons
df.loc[(df['region']=='South') & (df['profit']>0)]
```

```python
#without using loc we can also use filter condition
df[df['region']=='South']
```

```python
#inverse the condition
#give me all regions except South
df[~(df['region']=='South')]
```

```python

data_new={'name':['Sumit','Rahul','Ankit','Ankit'],
      'age':[25,30,30,30],
      'salary':[25000,30000,35000,35000],
      }

df=pd.DataFrame(data_new)
df

```

```python
# #it will drop all the duplicate rows keeping the first occurence (BY DEFAULT)
df.drop_duplicates()
```

```python
# #drop duplcuates keeping only LAST OCCURENCE
df.drop_duplicates(keep='last')
```

```python
#drop duplictes only keep first occurence
df.drop_duplicates(keep='first')
```

```python
#drop duplictes based on particular column
df.drop_duplicates(subset='age')
```

```python
df.drop_duplicates(subset='age',keep='last')
```

```python
df.drop_duplicates(subset='name')
```

```python
df.drop_duplicates(subset=['name','age'])
```
